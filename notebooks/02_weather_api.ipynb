```python
import sys
sys.path.append("..")

import pandas as pd
from sqlalchemy import text

from src.database import get_engine, load_weather
from src.weather import get_weather_data

engine = get_engine()

# Read the cities (with coordinates) from the database
with engine.connect() as conn:
    cities_df = pd.read_sql(
        text("SELECT city_name AS city, latitude, longitude FROM cities"), conn
    ).astype({"latitude": float, "longitude": float})

weather_df = get_weather_data(cities_df)
print(weather_df.dtypes)
display(weather_df.head())

load_weather(weather_df, engine)

# Verify
with engine.connect() as conn:
    print(pd.read_sql(text("SELECT COUNT(*) AS n FROM weather"), conn))
```
